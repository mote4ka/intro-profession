# Практика 1. Репозитории и интеграции

Поиск фильма через OMDb API. Выполните ячейки по порядку.
Ветка `master`: ключ вводится через скрытый ввод `getpass`.


```python
import json
import requests
import getpass
```

## Ввод данных

Введите название фильма, например `Batman`. Ключ получите на сайте OMDb и введите в скрытое поле.


```python
title = input("Введите фильм для поиска: ").strip()
api_key = getpass.getpass("Введите ваш OMDb API key: ").strip()
```

## HTTP-запрос и разбор JSON

Используем HTTPS, передаём параметры через `params`. `raise_for_status()` проверяет HTTP-статус; `json()` преобразует ответ в словарь.


```python
response = requests.get(
    "https://www.omdbapi.com/",
    params={"t": title, "apikey": api_key},
    timeout=15,
)
response.raise_for_status()
json_data = response.json()
print(json.dumps(json_data, ensure_ascii=False, indent=2))
```

## Нужные поля

Извлекаем название, год, жанр, режиссёра и рейтинг IMDb, а также актёров.


```python
movie_title = json_data.get("Title", "Нет данных")
year = json_data.get("Year", "Нет данных")
genre = json_data.get("Genre", "Нет данных")
director = json_data.get("Director", "Нет данных")
imdb_rating = json_data.get("imdbRating", "Нет данных")

print(f"Название: {movie_title}")
print(f"Год: {year}")
print(f"Жанр: {genre}")
print(f"Режиссёр: {director}")
print(f"Рейтинг IMDb: {imdb_rating}")
actors = json_data.get("Actors", "Нет данных")
print(f"Актёры: {actors}")
print(f"\nФильм «{movie_title}» ({year}) — жанр: {genre}. Режиссёр: {director}.")
```

## Если названия совпадают

Запрос с `t` возвращает одну запись. Чтобы выбрать конкретный фильм, можно добавить год (`y`) или указать IMDb ID (`i`). Для списка результатов используется `s`. См. [документацию OMDb](https://www.omdbapi.com/#parameters).

Перед коммитом убедитесь, что ключ отсутствует в коде, выводах и трассировках ошибок.
