# Практика 1. Репозитории и интеграции

Поиск фильма через OMDb API. Выполните ячейки по порядку.
Ветка `env-feature`: ключ читается из `.env` и не выводится на экран.


```python
import json
import requests
import os
from pathlib import Path
from dotenv import load_dotenv
from movie_api import MovieError, fetch_movie, format_movie
```

## Ввод данных

Введите название фильма, например `Batman`. Сначала создайте `.env` рядом с ноутбуком по образцу `.env.example`. Настоящая переменная окружения имеет приоритет над файлом.


```python
load_dotenv(Path.cwd() / ".env", override=False)
api_key = os.getenv("OMDB_API_KEY", "").strip()
title = input("Введите фильм для поиска: ").strip()
```

## HTTP-запрос и разбор JSON

Используем HTTPS, передаём параметры через `params`. Функция в `movie_api.py` проверяет HTTP-статус и поле `Response`, ограничивает время ожидания и выдаёт понятные сообщения без URL с ключом.


```python
movie = None
try:
    movie = fetch_movie(title, api_key)
except MovieError as error:
    print(f"Ошибка: {error}")
else:
    print(json.dumps(movie, ensure_ascii=False, indent=2))
```

## Нужные поля

Извлекаем название, год, жанр, режиссёра и рейтинг IMDb, а также актёров.


```python
if movie is not None:
    print(format_movie(movie))
```

## Если названия совпадают

Запрос с `t` возвращает одну запись. Чтобы выбрать конкретный фильм, можно добавить год (`y`) или указать IMDb ID (`i`). Для списка результатов используется `s`. См. [документацию OMDb](https://www.omdbapi.com/#parameters).

Перед коммитом убедитесь, что ключ отсутствует в коде, выводах и трассировках ошибок.
